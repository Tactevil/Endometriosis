```python
import zipfile

zip_filename = "UT-EndoMRI.zip"

try:
    with zipfile.ZipFile(zip_filename, 'r') as zip_ref:
        file_list = zip_ref.namelist()
        nii_files = [f for f in file_list if f.endswith('.nii.gz')]
        print(f"Total files in Zip: {len(file_list)}")
        print(f"Total NIfTI scans found: {len(nii_files)}")
        print("\nSample scan paths inside Zip:")
        for file in nii_files[:5]:
            print(f" - {file}")
            
        # Extract files properly
        zip_ref.extractall("./UT-EndoMRI_Extracted")
        print("\nExtraction complete into './UT-EndoMRI_Extracted'")
except FileNotFoundError:
    print(f"File {zip_filename} not found in current directory.")
```

```python
# ranaming the file
import os


# To rename the extracted directory:
if os.path.exists("UT-EndoMRI_Extracted"):
    os.rename("UT-EndoMRI_Extracted", "endometriosis")
    print("Renamed folder 'UT-EndoMRI_Extracted' to 'endometriosis'")
```
